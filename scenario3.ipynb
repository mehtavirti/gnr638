```python

import yaml
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

from models.load_models import load_model, freeze_backbone, unfreeze_last_block
from utils.dataset_loader import get_dataloaders
from utils.metrics import compute_accuracy, relative_performance_drop, train_val_gap

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
CONFIG_PATH = "configs/model_configs.yaml"

with open(CONFIG_PATH) as f:
    config = yaml.safe_load(f)

train_cfg = config["training"]
scenario_cfg = config["scenarios"]["few_shot"]

fractions = scenario_cfg["fractions"]

print("Fractions:", fractions)
```

```python
best_strategies = {
    "resnet50": "last_block",
    "densenet121": "last_block",
    "efficientnet_b0": "full_finetune"
}
```

```python
def train_model(model, train_loader, val_loader, epochs, lr,
                checkpoint_path=None, optimizer_path=None):

    criterion = nn.CrossEntropyLoss()

    optimizer = optim.Adam(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=lr,
        weight_decay=train_cfg["weight_decay"]
    )

    train_accs = []
    val_accs   = []
    start_epoch = 0
    best_val    = 0.0

    # RESUME: load optimizer state + accuracy history
    if optimizer_path and os.path.exists(optimizer_path):
        print(f"  [Resume] Loading resume state from {optimizer_path}")
        resume = torch.load(optimizer_path, map_location=device)
        optimizer.load_state_dict(resume["optimizer_state"])
        start_epoch = resume["epoch"]
        train_accs  = resume["train_accs"]
        val_accs    = resume["val_accs"]
        best_val    = resume["best_val"]
        print(f"  [Resume] Resuming from epoch {start_epoch+1}/{epochs}  (best val so far: {best_val:.4f})")

        if checkpoint_path and os.path.exists(checkpoint_path):
            model.load_state_dict(torch.load(checkpoint_path, map_location=device))
    else:
        print(f"  [Fresh] Starting training from epoch 1")

    for epoch in range(start_epoch, epochs):

        model.train()
        train_acc = 0

        for images, labels in tqdm(train_loader, leave=False):
            images = images.to(device)
            labels = labels.to(device)
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_acc += compute_accuracy(outputs, labels)

        train_acc /= len(train_loader)

        model.eval()
        val_acc = 0

        with torch.no_grad():
            for images, labels in val_loader:
                images = images.to(device)
                labels = labels.to(device)
                outputs = model(images)
                val_acc += compute_accuracy(outputs, labels)

        val_acc /= len(val_loader)

        train_accs.append(train_acc)
        val_accs.append(val_acc)

        print(f"  Epoch {epoch+1}/{epochs} | Train Acc {train_acc:.4f} | Val Acc {val_acc:.4f}", end="")

        # Save BEST checkpoint only when val improves
        if val_acc > best_val:
            best_val = val_acc
            if checkpoint_path:
                torch.save(model.state_dict(), checkpoint_path)
                print(f"  --> best saved (val={best_val:.4f})", end="")

        print()

        # Save RESUME state after every epoch
        if optimizer_path:
            torch.save({
                "epoch"          : epoch + 1,
                "optimizer_state": optimizer.state_dict(),
                "train_accs"     : train_accs,
                "val_accs"       : val_accs,
                "best_val"       : best_val,
            }, optimizer_path)

    return train_accs, val_accs
```

```python
import os

SAVE_DIR = "/content/drive/MyDrive/gnr638_scenario3_models"
os.makedirs(SAVE_DIR, exist_ok=True)

RESULTS_FILE = os.path.join(SAVE_DIR, "scenario3_results.csv")

# Load existing results so completed (model, frac) pairs are skipped
if os.path.exists(RESULTS_FILE):
    results = pd.read_csv(RESULTS_FILE).to_dict("records")
    print(f"Loaded {len(results)} existing results from Drive.")
else:
    results = []

def already_done(model_name, frac):
    return any(r["model"] == model_name and abs(r["fraction"] - frac) < 1e-6 for r in results)

for model_name, strategy in best_strategies.items():

    print("\n=====================")
    print("Model:", model_name)

    for frac in fractions:

        print(f"\nFraction: {frac}")

        best_ckpt_path   = os.path.join(SAVE_DIR, f"{model_name}_fewshot_{frac}_best.pth")
        resume_ckpt_path = os.path.join(SAVE_DIR, f"{model_name}_fewshot_{frac}_resume.pth")

        if already_done(model_name, frac):
            print("  [Skip] Already completed.")
            continue

        train_loader, val_loader, classes = get_dataloaders(
            DATA_DIR,
            batch_size=train_cfg["batch_size"],
            num_workers=train_cfg["num_workers"],
            seed=train_cfg["seed"],
            img_size=train_cfg["img_size"],
            fraction=frac
        )

        num_classes = len(classes)
        model = load_model(model_name, num_classes).to(device)

        if strategy == "last_block":
            model = unfreeze_last_block(model, model_name)
        elif strategy == "full_finetune":
            for p in model.parameters():
                p.requires_grad = True

        # Load best weights before resuming so model is in the right state
        if os.path.exists(resume_ckpt_path) and os.path.exists(best_ckpt_path):
            model.load_state_dict(torch.load(best_ckpt_path, map_location=device))

        train_accs, val_accs = train_model(
            model,
            train_loader,
            val_loader,
            scenario_cfg["epochs"],
            train_cfg["lr"],
            checkpoint_path=best_ckpt_path,
            optimizer_path=resume_ckpt_path
        )

        best_val = max(val_accs)

        results.append({"model": model_name, "fraction": frac, "best_val_acc": best_val})

        results_df = pd.DataFrame(results)
        results_df.to_csv(RESULTS_FILE, index=False)
        print(f"  Results saved to: {RESULTS_FILE}")

        # Remove resume checkpoint now that this run is fully done
        if os.path.exists(resume_ckpt_path):
            os.remove(resume_ckpt_path)
            print(f"  Resume checkpoint cleaned up.")
```

```python
from utils.metrics import relative_performance_drop, train_val_gap

metrics = []

for model in ["resnet50", "densenet121", "efficientnet_b0"]:

    subset = results_df[results_df["model"] == model]

    acc100 = subset[subset["fraction"] == 1.0 ]["best_val_acc"].values[0]
    acc20  = subset[subset["fraction"] == 0.2 ]["best_val_acc"].values[0]
    acc5   = subset[subset["fraction"] == 0.05]["best_val_acc"].values[0]
    drop   = relative_performance_drop(acc100, acc5)

    gaps = {}
    for frac in [1.0, 0.2, 0.05]:
        c = all_curves[(model, frac)]
        gaps[frac] = train_val_gap(c["train"][-1], c["val"][-1])

    metrics.append({
        "model"         : model,
        "acc_100pct"    : round(acc100,     4),
        "acc_20pct"     : round(acc20,      4),
        "acc_5pct"      : round(acc5,       4),
        "relative_drop" : round(drop,       4),
        "gap_100pct"    : round(gaps[1.0],  4),
        "gap_20pct"     : round(gaps[0.2],  4),
        "gap_5pct"      : round(gaps[0.05], 4),
    })

metrics_df = pd.DataFrame(metrics)
metrics_df.to_csv(os.path.join(SAVE_DIR, "scenario3_metrics.csv"), index=False)
print(metrics_df)
```
